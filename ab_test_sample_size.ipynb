```python
from statsmodels.stats.power import NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize

def n_per_arm(baseline, target, alpha=0.05, power=0.8):
    es = proportion_effectsize(target, baseline)
    return NormalIndPower().solve_power(effect_size=es, alpha=alpha,
                                        power=power, ratio=1, alternative='two-sided')

print(round(n_per_arm(0.1724, 0.1424)))  # low reviews, 3-point reduction
print(round(n_per_arm(0.0100, 0.0150)))  # repeat rate, +0.5 points
print(round(n_per_arm(0.0100, 0.0130)))  # repeat rate, +0.3 points
```

```python
# Assumptions: alpha = 0.05 (two-sided), power = 0.80, equal split between arms.
# Baselines measured from order_base for the nine high-late states.
# Effect sizes are assumed (minimum improvement worth detecting), not observed.
```
